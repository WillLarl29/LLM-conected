## Creación del ambiente virtual

Ya hemos trabajado con la parte de indexación, ahora nos enfocaremos en la construcción de las queries, lo cual es más avanzado, incluyendo el retrieval (recuperación). Utilizaremos VS Code, donde hemos creado un directorio vacío llamado "LangChain RAG".

Procederemos a abrir el terminal en VS Code, seleccionando "Nuevo Terminal" y eligiendo Command Prompt en lugar de PowerShell, ya que es más adecuado para nuestro caso. Lo primero es verificar la versión de Python instalada. Recomendamos tener la versión Python 3.13.2, ya que versiones más antiguas o más nuevas podrían no funcionar correctamente.

Para verificar la versión de Python, utilizamos el siguiente comando:

```python
python -V
```

No nos detendremos en cómo instalar Python, ya que asumimos que quienes están en este curso ya saben hacerlo. Hay otros entrenamientos en Alura que muestran cómo instalar Python. Una vez instalada la versión adecuada, crearemos un ambiente virtual con el siguiente comando:

```python
python -m venv .venv
```

Esto crea un directorio para el ambiente virtual, que debemos activar con el comando:



```python
.\.venv\Scripts\activate
```

## Configuración del entorno y carga de variables

De esta manera, ya tenemos nuestros modelos en nuestro ambiente virtual. El siguiente paso es crear un archivo con nuestras variables. Debemos asegurarnos de estar fuera de cualquier directorio específico y crear un archivo llamado .env. Este archivo contendrá todas las variables de entorno necesarias.

Luego, crearemos un archivo llamado rag.py, donde trabajaremos efectivamente. Vamos a copiar rápidamente nuestras variables de entorno, asegurándonos de que no queden expuestas. El siguiente paso es comenzar a importar nuestros paquetes. Importamos os y from dotenv import load_dotenv para cargar las variables de ambiente. También cargaremos nuestro directory loader.

Estos son comandos que ya hemos visto en pasos anteriores, ya que vamos a realizar todo nuestro pipeline de rack. Repetiremos algunos pasos, haciendo bastante copy-paste de lo que ya hemos trabajado. Traeremos también las bibliotecas de transformers y tokenizer. Cargaremos nuestras variables de entorno y las traeremos aquí.

Para cargar las variables de entorno, utilizamos el siguiente código:

```python
import os
from dotenv import load_dotenv

load_dotenv()
GEMINI_API_KEY = os.getenv("GEMINI_API_KEY")
LANGSMITH_TRACING = os.getenv("LANGSMITH_TRACING")
LANGSMITH_API_KEY = os.getenv("LANGSMITH_API_KEY")
```
